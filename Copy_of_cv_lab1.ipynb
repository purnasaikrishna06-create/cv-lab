```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
def show_image(title,img,cmap=None):
  plt.figure(figsize=(5,5))
  plt.title(title)
  if cmap:
    plt.imshow(img,cmap=cmap)
  else:
      plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
      plt.axis('off')
      plt.show()
```

```python
img1=cv2.imread("/content/2.jpeg")
img2 = cv2.imread('/content/images.jpg')
```

```python
img1.shape
(350,424,3)
```

```python
img2=cv2.resize(img2,(img1.shape[1],img1.shape[0]))
```

```python

resized_img = cv2.resize(img1,(300,300))
```

```python
cropped_img=img1[50:200,100:50]
```

```python
gray_img = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
show_image("Grayscale Image", gray_img, cmap='gray')
```

```python
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))
added_img = cv2.addWeighted(img1,0.5,img2,0.5,0)
show_image("Arithmetic Addition(Blending)",added_img)
```

```python
subtracted_img = cv2.subtract(img1,img2)
show_image("Arithmetic Subtraction",subtracted_img)
```

```python
rectangle = np.zeros((img1.shape[0], img1.shape[1]), dtype="uint8")
cv2.rectangle(rectangle, (50, 50), (300, 300), 255, -1)
```

```python
circle = np.zeros((img1.shape[0], img1.shape[1]), dtype="uint8")
cv2.circle(circle, (200, 200), 100, 255, -1)
```

```python
bitwise_and = cv2.bitwise_and(rectangle, circle)
show_image("Bitwise AND", bitwise_and, cmap='gray')
```

```python
bitwise_or = cv2.bitwise_or(rectangle, circle)
show_image("Bitwise OR", bitwise_or, cmap='gray')
```

```python
bitwise_xor = cv2.bitwise_xor(rectangle, circle)
show_image("Bitwise XOR", bitwise_xor, cmap='gray')
```

```python
bitwise_not = cv2.bitwise_not(circle)
show_image("Bitwise NOT", bitwise_not, cmap='gray')
```
